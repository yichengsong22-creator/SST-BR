```python
import matplotlib.pyplot as plt
from matplotlib.patches import FancyBboxPatch

fig, ax = plt.subplots(figsize=(12, 4.5))
boxes = [
    (0.02, 0.58, 0.16, 0.22, "Radar windows\nΔt", "#d9edf7"),
    (0.25, 0.58, 0.18, 0.22, "Spatial + spectral\nlocal representation", "#cfe8f3"),
    (0.52, 0.70, 0.18, 0.18, "Temporal fusion\nover T", "#dff2e1"),
    (0.77, 0.70, 0.18, 0.18, "Baseline branch", "#dff2e1"),
    (0.52, 0.28, 0.18, 0.18, "Current window", "#d9edf7"),
    (0.77, 0.28, 0.18, 0.18, "Residual branch", "#d9edf7"),
]
for x, y, w, h, label, color in boxes:
    ax.add_patch(FancyBboxPatch((x, y), w, h, boxstyle="round,pad=0.02", fc=color, ec="#19466e", lw=1.4))
    ax.text(x+w/2, y+h/2, label, ha="center", va="center", fontsize=10)
for start, end, color in [((0.18,0.69),(0.25,0.69),"#247ba0"),((0.43,0.69),(0.52,0.79),"#2e8b57"),
                          ((0.70,0.79),(0.77,0.79),"#2e8b57"),((0.43,0.64),(0.52,0.37),"#247ba0"),
                          ((0.70,0.37),(0.77,0.37),"#247ba0")]:
    ax.annotate("", xy=end, xytext=start, arrowprops={"arrowstyle":"->","lw":2,"color":color})
ax.text(0.87, 0.57, "baseline + residual → HR", ha="center", weight="bold", color="#19466e")
ax.set(xlim=(0,1), ylim=(0,1), title="SST-BR data flow")
ax.axis("off")
plt.show()
```

## Temporal organization

An observation of duration **T** is divided into **N** local windows of duration **Δt**. The green path summarizes information over T for the baseline; the blue path preserves current-window information for the residual.

## Baseline–residual interpretation

The baseline describes the slower subject context. The residual describes the current local departure. Their additive combination produces HR.
