# SST-BR Interactive Demo

Full executable release

## Method overview

`local window Δt → spatial–spectral representation`  
`multiple local representations over T → temporal fusion → baseline`  
`current local representation → residual`  
`baseline + residual → HR`

```python
import matplotlib.pyplot as plt
import numpy as np
from sstbr.demo.synthetic import generate_synthetic_recording

recording = generate_synthetic_recording()
fig, axes = plt.subplots(1, 2, figsize=(11, 3.2))
axes[0].imshow(np.abs(recording.radar_windows[0]), aspect="auto", origin="lower", cmap="Blues")
axes[0].set(title="Synthetic radar magnitude", xlabel="Slow-time sample", ylabel="Range bin")
t_ppg = np.arange(len(recording.ppg)) / recording.ppg_hz
axes[1].plot(t_ppg, recording.ppg, color="#d95f3d", lw=1.2)
axes[1].set(title="Synthetic PPG", xlabel="Time (s)", ylabel="Amplitude")
fig.tight_layout()
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(10, 2.2))
colors = ["#4aa8d8" if i % 2 == 0 else "#70b77e" for i in range(6)]
for i, color in enumerate(colors):
    ax.barh(0, 1, left=i, color=color, edgecolor="white", height=0.55)
    ax.text(i + 0.5, 0, f"W{i+1}", ha="center", va="center", color="white", weight="bold")
ax.annotate("local path (Δt)", xy=(0.5, 0.38), xytext=(0.5, 0.78), ha="center",
            arrowprops={"arrowstyle": "->", "color": "#247ba0"}, color="#247ba0")
ax.annotate("temporal path (T)", xy=(3.0, -0.38), xytext=(3.0, -0.78), ha="center",
            arrowprops={"arrowstyle": "-[,widthB=4.7", "color": "#2e8b57"}, color="#2e8b57")
ax.set(xlim=(0, 6), ylim=(-1, 1), title="T divided into N local windows of duration Δt")
ax.axis("off")
plt.show()
```

## Local representation

This cell calls the complete spatial/spectral extractor.

```python
from sstbr import SSTBRCore
core = SSTBRCore(radar_fs=recording.radar_hz)
representation = core.extract_local_representation(recording.radar_windows[0], recording.range_grid)
print("Local representation shape:", representation.shape)
print("Finite values:", int(np.isfinite(representation).sum()), "/", len(representation))
```

```python
from sstbr.demo.synthetic import run_demo
demo_result = run_demo()
print(f"baseline = {demo_result['baseline']:.3f} BPM")
print(f"residual = {demo_result['residual']:.3f} BPM")
print(f"final = {demo_result['final']:.3f} BPM")

metrics = demo_result['metrics']
print('10-s metrics:', {k: round(v, 3) if isinstance(v, float) else v for k, v in metrics.items()})
```
